<h1 style="font-size: 250%; font-family: cursive;"><b>Singular Value Decomposition: Theory and its Applications</b></h1>


<p style="font-family:cursive;"><b> Singular Value Decomposition (SVD) </b> is a powerful concept in linear algebra that plays a crucial role in understanding and manipulating matrices.It allows any given matrix to be decomposed into three distinct matrices which can be expressed mathematically as:
</p>

$$ A = U \Sigma V^T $$ 

$where:$
- $A$ is any $m \times n$ matrix
- $U$ is an $m \times m$ orthogonal matrix where the columns are the eigenvectors of $AA^T$ also known as the left singular vectors of $A$.
- $\Sigma$ is an $m \times n$ diagonal matrix with non-negative real numbers on the diagonal known as the singular values of $A$.
- $V$ is an $n \times n$ orthogonal matrix where the columns are the eigenvectors of $A^TA$ also known as the right singular vectors of $A$.

<p style='font-family:cursive;'>SVD reveals that any matrix—regardless of its symmetry, rank, or shape—can be decomposed into three special matrices. This decomposition is a cornerstone of data analysis, offering insights that extend far beyond pure mathematics.</p>


<h2 style="font-size: 120%; color:tomato; font-family: cursive;"><b>The power of Rectangular Matrices</b></h2>

SVD isn’t just for square matrices; it extends beautifully to rectangular ones, which are commonly used in real-world problems. This generalization is critical, as it allows SVD to handle matrices where the number of rows and columns are not the same. Visualizing this requires understanding the difference between vectors in different dimensions, say from $R^2$ to $R^3$.<br>
A key function of matrices, especially rectangular ones, }is their ability to transform vectors from one space to another. For example, a 2x3 matrix can map a vector from a 3-dimensional space to a 2-dimensional space. This transformation is crucial in many machine learning algorithms, where data from higher dimensions must be projected into lower-dimensional spaces without losing important features.

<h2 style="font-size: 120%; color:tomato; font-family: cursive;"><b>Step by Step Transformation</b></h2>
SVD allows us to break down a matrix's transformation into three distinct operations:

1. **Rotation**: The matrix $V^T$ applies the rotation such that the right singular vector returns to standard basis.

2. **Scaling**: The matrix $\Sigma$ scales the rotated vector by the singular values.

3. **Rotation**: Finally, the matrix $U$ will revert step 1, which means it rotates the standard basis to align with the left singular vectors.



<h2 style="font-size:120%; color:tomato; font-family: cursive;"><b>SVD using rank-1 matrices</b></h2>

We now know that we can write any matrix as : $ A = U \Sigma V^T $ 

Another interesting way to represent same decomposion is :
$$ A= U \sigma V^T =\vec{ u_1 } \sigma_1 \vec{v_1}^T + \vec{ u_2 } \sigma_2 \vec{v_2}^T + . . . + \vec{ u_n } \sigma_n \vec{v_n}^T  $$

where, 
- $ \sigma_i $    are the singular values (diagonal elements of $ \Sigma$).
- $ \vec{u_i} $   are the columns of U (left singular vectors).
- $ \vec{v_i}^T $ are the rows of $ V^T $ (right singular vectors).

Now, as we know, the singular values $( \sigma )$ are arranged in descending order in the diagonal matrix $( \Sigma )$. This means that the earlier terms in the decomposition:

$[ A = \sigma_1 \vec{u_1} \vec{v_1}^T + \sigma_2 \vec{u_2} \vec{v_2}^T + \dots + \sigma_n \vec{u_n} \vec{v_n}^T ]$

will have larger singular values $( \sigma )$, while the later terms will have smaller singular values or even zero. 

This gives us an important insight: we can truncate the decomposition by keeping only the terms with the largest singular values and discard the later terms with smaller or zero singular values. By doing this, we can still obtain a good approximation of the original matrix $ A $.

For example, if we keep only the first $( k )$ terms where $( k < n )$, the approximate matrix $( A_k )$ can be written as:

$[ A_k = \sigma_1 \vec{u_1} \vec{v_1}^T + \sigma_2 \vec{u_2} \vec{v_2}^T + \dots + \sigma_k \vec{u_k} \vec{v_k}^T ]$

Here, $( A_k )$ is a rank-$( k )$ approximation of the original matrix $( A )$. This approximation is especially useful in applications such as data compression, noise reduction, and dimensionality reduction, where we can represent the original matrix using only a few significant singular values and vectors.




<h2 style="font-size:120%; color:tomato; font-family: cursive;"><b>Applications of SVD</b></h2>

1. **Image Compression**: SVD is widely used in image compression, where images are represented as matrices. By keeping only the significant singular values and vectors, we can compress the image while retaining most of the important features.

2. **Collaborative Filtering**: SVD is used in recommendation systems to predict user preferences. By decomposing the user-item matrix, we can identify latent factors that influence user preferences and make accurate recommendations.

3. **Dimensionality Reduction**: SVD is used in machine learning for dimensionality reduction. By keeping only the significant singular values and vectors, we can reduce the number of features while retaining most of the important information.

4. **Signal Processing**: SVD is used in signal processing to denoise signals. By decomposing the signal matrix, we can remove noise and extract the underlying signal.

#### Visualizing an Image numerically as a matrix

```python
from skimage.io import imread
from skimage.transform import resize
import matplotlib.pyplot as plt
import numpy as np
```

```python
img = imread('images/cow.jpg', as_gray=True)
data = resize(img, (12,12)) #for simpler visualization
data = np.round(data,2)
```

```python
plt.figure(figsize=(20,20))
plt.subplot(131)
plt.imshow(img, cmap='gray')
plt.title(f'Orginal Image Grayscaled [shape:{img.shape}]')

plt.subplot(132)
plt.imshow(data, cmap='gray')
plt.title(f'Resized Image [shape;{data.shape}]')

plt.show()
```

```python
from sympy import Matrix, pprint

#using previouly resized matrix
A = Matrix(data)

# Pretty print the matrix
pprint(A)
```

#### Load image and apply SVD

```python
#using previously loaded image i.e img. 

# Perform SVD
U, S, Vt = np.linalg.svd(img, full_matrices=False)

# Convert singular values to a diagonal matrix
Sigma = np.diag(S)
```

#### Truncate SVD for Compression

```python
# Choose the number of singular values to keep
k_values = [5, 25, 50,200, 408]


plt.figure(figsize=(15, 8))
plt.subplot(2, 3, 1)
plt.imshow(img, cmap='gray')
plt.title('Original Image')
plt.axis('off')

# Compress and display images for different k values
for i, k in enumerate(k_values):
    # Truncate U, Sigma, and Vt
    U_k = U[:, :k]
    Sigma_k = Sigma[:k, :k]
    Vt_k = Vt[:k, :]

    # Reconstruct the compressed image
    compressed_image = U_k @ Sigma_k @ Vt_k

    # Display the compressed image
    plt.subplot(2, 3, i + 2)
    plt.imshow(compressed_image, cmap='gray')
    plt.title(f'k = {k}')
    plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
k=50 
U_k = U[:, :k]
Sigma_k = Sigma[:k, :k]
Vt_k = Vt[:k, :]

print("Shape of U, sigma, Vt before compression   :  ",U.shape, Sigma.shape, Vt.shape)
print("Shape of U, sigma, Vt after trucation by k :  ",U_k.shape, Sigma_k.shape, Vt_k.shape)
print("shape of compressed image : ",compressed_image.shape)
```

You might notice that the compressed image appears to be the same size as the original image. This is because we reconstructed the image using the truncated matrices.

### So, how is the compression actually happening?

The key to understanding the compression lies in the number of values (or "pixels") required to represent the original image compared to the truncated matrices. We can calculate this by comparing the number of elements in the original image matrix to the total number of elements in the truncated $ U_k ,  \Sigma_k , and  V_k^T $ matrices.

This comparison highlights the reduction in storage requirements achieved through the SVD-based compression as we will be only storing the significant singular values and vectors, which are much smaller in number compared to the original image matrix.


```python

# Number of pixels in the original image
num_pixels_original = img.size

# Number of pixels in the compressed image
num_pixels_compressed = U_k.size + Sigma_k.size + Vt_k.size

print("Number of pixels in the original image: ", num_pixels_original)
print("Number of pixels in the compressed image: ", num_pixels_compressed)

# Calculate the compression ratio

compression_ratio = num_pixels_compressed / num_pixels_original
print(f"Compression ratio: {compression_ratio:.4f}")
print("The compressed image is only {:.2f}% of the size of the original image.".format(compression_ratio * 100))
```

Same process also can be applied for RGB images by applying SVD to each channel separately.
